```python
def prime(num):
    count=0
    i=1
    if num==1:
        return f"{num} -> not a prime number"
    while i<=num:
        if num%i==0:
            count+=1
        i+=1
        if count>2:
            return f"{num} -> it is not a prime number"
    if count==2:
        return f"{num} -> it is prime number"
n=int(input("enter the number to check the prime number"))
print(prime(n))
```

# 7 GCD

## 7.1 WAP to display the HCF/GCD of the guven numbers

```python
def gcd_(n,m):
    lower=n
    gcd=0
    loopc=0
    if m<n:
        lower=m
    for i in range(1,lower+1):
        loopc+=1
        if n%i==0 and m%i==0:
            gcd=i
    return f"the gcd of {n} and {m} is {gcd} \n with loop count {loopc}"
n=int(input("enter the number 1 to check the hcf: "))
m=int(input("enter the number 2 to check the hcf: "))
print(gcd_(n,m))
```

## 7.2 to check the HCF using the reverse

```python
def gcd_(n,m):
    lower=n
    gcd=1
    loopc=0
    if m<n:
        lower=m
    for i in range(lower,0,-1):
        loopc+=1
        if n%i==0 and m%i==0:
            gcd=i
            return f"the gcd of {n} and {m} is {gcd} \n with loop count {loopc}"
    
n=int(input("enter the number 1 to check the hcf: "))
m=int(input("enter the number 2 to check the hcf: "))
print(gcd_(n,m))
```

## 8.1 Fibonacci series

```python
def fibb(n):
    n1,n2=0,1
    while n>0:
        print(n1,end=" ")
        add=n1+n2
        n1=n2
        n2=add
        n-=1
    return ""
n=int(input("enter the range to get the series: "))
print(fibb(n))
```

```python
def fibb(n):
    n1,n2=0,1
    l=[]
    while n>0:
        print(n1,end=" ")
        l.insert(0,n1)
        add=n1+n2
        n1=n2
        n2=add
        n-=1
    print()
    return l 
n=int(input("enter the range to get the series: "))
print(fibb(n))
```

```python
def lcm(n1,n2):
    lcm=n1
    if n1<n2:
        lcm=n2
    while lcm<=(n1*n2):
        if lcm%n1==0 and lcm%n2==0:
            print(f"the LCM of {n1} and {n2} is: ",end="")
            return lcm
        lcm+=1
n1=int(input("enter the number"))
n2=int(input("enter the number"))
print(lcm(n1,n2))
```

## 10.10 Prime no. using recurssion
