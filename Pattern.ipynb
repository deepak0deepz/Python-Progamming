## 1 ▶︎ Pattern of * 

```python
n=int(input("enter no : "))
for i in range(1,n+1):
    print("* "*i," "*(n-i))
for i in range(n-1,0,-1):
    print("* "*i," "*(n-i))
```

```python
n=int(input("enter no : "))
x=1
for i in range(1,n*2):
    for j in range (1,x+1):
        print("*",end=" ")
    print()
    if i<n:
        x+=1
    else:
        x-=1
```
